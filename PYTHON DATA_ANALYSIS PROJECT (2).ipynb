```python
# total transaction count by gender wise in the bar
ax=sns.countplot(x="Gender",data=df)
for i in ax.containers:
    ax.bar_label(i)
plt.title("Gender Distribution",fontsize=10)    
plt.xlabel("GENDER",fontsize=10,color="k")
plt.ylabel("COUNT",fontsize=15,color="r")
```

```python
# total_count by Gender_wise in pie chart
Gender_count=df['Gender'].value_counts()
print(Gender_count)
```

```python
# total transaction count by gender wise in pie chart
Gender_count=df['Gender'].value_counts()
A = [0.1, 0]
plt.pie(Gender_count,labels=Gender_count,autopct="%1.2f%%",startangle=90,explode=A)
plt.title("Gender Distribution", fontsize=20)

```

```python
df.groupby("Product_Category")['Amount'].sum()
```

```python
df.groupby("Gender",as_index=False)['Amount'].sum()
```

```python
# gender wise total sales
gender_wise=df.groupby("Gender",as_index=False)['Amount'].sum().sort_values(by ="Amount",ascending=False)
```

```python
ax=sns.barplot(x='Gender',y="Amount",data=gender_wise)
for i in ax.containers:
       ax.bar_label(i,fmt='%F')
plt.title('Gender_wise Total Sales Amount', fontsize=15)
```

## Age Group wise total transactions



```python
age_group=sns.countplot(x='Age Group',data=df)
for i in age_group.containers:
    age_group.bar_label(i)
```

## age group and gender wise transaction count

```python
group=sns.countplot(x="Age Group",hue='Gender',data=df)
```

# # age group wise transaction count(sorting)


```python
age_group=df["Age Group"].value_counts().sort_values(ascending=True)
A=age_group.index
sns.countplot(data=df,x='Age Group',order=A)
```

## age group wise total Amount

```python
age_group=df.groupby('Age Group',as_index=False)["Amount"].sum().sort_values(by="Amount", ascending=False)
ax=sns.barplot(x='Age Group',y="Amount",data =age_group)
for i in ax.containers:
    ax.bar_label(i,fmt='%i')
```

## state wise Analysis

```python
# orders wise top 5 state
state_wise=df.groupby('State',as_index=False)['Orders'].sum().sort_values(by = "Orders", ascending=True)
plt.figure(figsize=(20,10))
plt.title("State Wise Analysis",fontsize=25)
ax=sns.barplot(y='Orders',x="State",data=state_wise)
for i in ax.containers:
    ax.bar_label(i)
```

## Amount wise top 5 state

```python
Amount_wise=df.groupby('State',as_index=False)['Amount'].sum().sort_values(by="Amount",ascending=False)
plt.figure(figsize=(30,19))
plt.title("Amount wise top 5 State",fontsize=30,color="r")
ax=sns.barplot(x='State',y='Amount',data=Amount_wise)
for i in ax.containers:
    ax.bar_label(i,fmt="%i")
```

# Product Category Analysis

```python
# Product_Category Wise Transactions Count
product_category=df['Product_Category'].value_counts().sort_values(ascending=False)
Orderss=product_category.index
ax=sns.countplot(data=df,x='Product_Category',order=Orderss)
for i in ax.containers:
    ax.bar_label(i)
```

# #Amount wise Product Category

```python
Amount_wise =df.groupby('Product_Category',as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
plt.figure(figsize=(19,10))
plt.title("Amount wise Product Category",fontsize=15)
ax=sns.barplot(data=Amount_wise,x='Product_Category',y='Amount')
for i in ax.containers:
    ax.bar_label(i,fmt="%i")

```

## product category and gender_wise Transactions count

```python
gender_wise =df.groupby(['Product_Category','Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
plt.title("Product_Category and Gender_wise Transactions",fontsize=10)
ax=sns.barplot(x='Product_Category',y='Amount',data=gender_wise,hue='Gender')
for i in ax.containers:
    ax.bar_label(i,fmt="%i")
```

## profession wise Analysis

```python
plt.figure(figsize=(20,15))
sns.countplot(data=df,x="Profession")

```

```python
professions_count=df['Profession'].value_counts().sort_values(ascending=False)
A=professions_count.index
plt.figure(figsize=(19,15))
plt.title("Professions wise Transactions Counts",fontsize=19)
ax=sns.countplot(data=df,x='Profession',order=A)
for i in ax.containers:
    ax.bar_label(i)
```

## Amount wise top Professions

```python
Amount_wise=df.groupby('Profession',as_index=False)['Amount'].sum().sort_values(by="Amount",ascending=False)
plt.figure(figsize=(30,19))
plt.title("Amount wise top 5 Profession",fontsize=30,color="r")
ax=sns.barplot(x='Profession',y='Amount',data=Amount_wise)
for i in ax.containers:
    ax.bar_label(i,fmt="%i")
```

## order wise top 10 product_id


```python
order_wise=df.groupby('Product_ID',as_index=False)['Orders'].sum().sort_values(by = "Orders", ascending=False).head(10)
plt.figure(figsize=(20,10))
plt.title("Orders Wise Analysis",fontsize=25)
ax=sns.barplot(y='Orders',x="Product_ID",data=order_wise)
for i in ax.containers:
    ax.bar_label(i)
```
